```python
def generate_arrivals(seed, lam, T_max):
    rng = np.random.default_rng(seed)

    t = 0.0
    arrivals = []

    while t <= T_max:
        t += -np.log(1 - rng.random()) / lam
        if t < T_max:
            arrivals.append(t)

    return arrivals
```

```python
def build_Nend(arrivals, Delta, num_steps):
    Nend = []
    for ts in range(num_steps):
        upper = (ts + 1) * Delta
        count = sum(a <= upper for a in arrivals)
        Nend.append(count - 1)  # last index
    return Nend
   # print(Nend)
```

```python
def segment(x):
    if x >= S["Exit_Distance"].iloc[-1]:
        return num_segments
    for i in range(num_segments):
        if x < S.loc[i]["Exit_Distance"]:
            return i
    return -1
```

```python
def compute_density(count):
  dens=[0]*(num_segments +1)
  vrhol=[0]*(num_segments +1)

  for i in range(num_segments):
    capacity = S.loc[i]["Car_Capacity"]
    if capacity > 0:
      dens[i] = count[i] / capacity
    else:
      dens[i] = 0
    vrhol[i] = vmax*(1 - dens[i])

  return dens, vrhol
```

```python
def vel_density(vrhol):
  for car in range(num_cars):
    vrhoc[car]=vrhol[carloc[car][ts]]
  return vrhoc
```

```python
def update_car(dt, a, x_prev, v_prev, vrho_prev):
    vrho_prev = max(vrho_prev, 0.5)  # minimum crawl speed, avoids permanent freeze
    x = x_prev + v_prev * dt + 0.5 * a * (1 - (v_prev / vrho_prev) ** 2) * dt ** 2
    v = (x - x_prev) / dt

    if x < x_prev:      # never let a car move backward
        x = x_prev
        v = 0.0

    return x, v
```

```python
def simul(numlanes, lam):
  all_arrivals = []
  for seed in range(numlanes):

      arrivals = generate_arrivals(
          seed=seed,
          lam=lam,
          T_max=T_max
      )

      all_arrivals.extend(arrivals)

  num_cars = len(all_arrivals)

  # sort the combined arrivals
  all_arrivals.sort()

  Nend = build_Nend(all_arrivals, Delta, T_steps)

  return num_cars, all_arrivals, Nend
```

```python
import numpy as np
import pandas as pd

lam = [0.5, 0.45, 0.4, 0.35, 0.3, 0.25]
numlanes = [17, 4, 4, 4, 4, 4]
T_max = 900  # 15 minutes -- long enough for cars to actually traverse the corridor
Delta = 5
T_steps = int(T_max/Delta)
a = 1.5  # m/s^2 -- realistic car acceleration (0.0925 was unrealistically low, cars barely moved)
vmax = 30

S = pd.DataFrame({
    "Exit_Distance":[122, 222, 347, 447, 547, 647, 747, 847, 947, 1047, 1147, 1247, 1347, 1447, 1547, 1677, 1777, 1877, 1977, 2077, 2177, 2277, 2377, 2477, 2577, 2677, 2777, 2877, 2977, 3077, 3177, 3277, 3377, 3477, 3577, 3677, 3777, 3877, 3977, 4077, 4177, 4277, 4377, 4477, 4577, 4677, 4777, 4877, 4977, 5077
],
    "Car_Capacity": [
414, 60, 75, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 104, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80, 80
]
})

i_arrivals = [[] for _ in range(len(numlanes))]
i_numcars = [None for _ in range(len(numlanes))]
i_Nend = [None for _ in range(len(numlanes))]
i_start = [0, 1677, 3577, 8527, 12177, 14577]

# Extend segment table so it covers all interchanges, not just up to 5077m
last_known = S["Exit_Distance"].iloc[-1]
target_end = max(i_start) + 500
extra_distances = list(range(int(last_known) + 100, int(target_end) + 100, 100))
extra_capacity = [80] * len(extra_distances)

S = pd.concat([
    S,
    pd.DataFrame({"Exit_Distance": extra_distances, "Car_Capacity": extra_capacity})
], ignore_index=True)

num_segments = len(S)

##########################################################
for inter in range(len(numlanes)):
  i_numcars[inter], i_arrivals[inter], i_Nend[inter] = simul(numlanes[inter], lam[inter])
  print(f'Number of cars of {inter} = ', i_Nend[inter])
  input("Press ENTER to continue")
```

```python
##########################################################
# Combine all arrivals from all interchanges
##########################################################


all_arrivals = []
all_start = []
all_interchange = []

for inter in range(len(numlanes)):

    for arrival in i_arrivals[inter]:

        all_arrivals.append(arrival)
        all_start.append(i_start[inter])
        all_interchange.append(inter)

##########################################################
# Sort cars by arrival time
##########################################################

cars = list(zip(all_arrivals, all_start, all_interchange))

cars.sort(key=lambda x: x[0])

all_arrivals = [car[0] for car in cars]
all_start = [car[1] for car in cars]
all_interchange = [car[2] for car in cars]

num_cars = len(all_arrivals)

print("Total cars =", num_cars)
```

```python
##########################################################
# Global Nend
##########################################################

Nend = []

for ts in range(T_steps):

    current_time = (ts + 1) * Delta

    last_car = -1

    for car in range(num_cars):

        if all_arrivals[car] <= current_time:
            last_car = car
        else:
            break

    Nend.append(last_car)

print(Nend)
```

```python
##########################################################
# Initialize simulation arrays
##########################################################

x = [[all_start[car]] * T_steps for car in range(num_cars)]

v = [[0.0] * T_steps for _ in range(num_cars)]

vrhoc = [[0.0] * T_steps for _ in range(num_cars)]

carloc = [[-1] * T_steps for _ in range(num_cars)]

loc = [[] for _ in range(T_steps)]

count = [[0] * num_segments for _ in range(T_steps)]

density = [[0] * num_segments for _ in range(T_steps)]

vrhol = [[0] * num_segments for _ in range(T_steps)]

xloc = [[] for _ in range(T_steps)]

print("\nMerged car list:\n")

print("Car\tArrival\tStart\tInterchange")

for car in range(min(20, num_cars)):
    print(
        f"{car}\t"
        f"{all_arrivals[car]:7.2f}\t"
        f"{all_start[car]:6.0f}\t"
        f"{all_interchange[car]}"
    )
```

```python
##########################################################
# Run Simulation
##########################################################
verbose = True

for ts in range(T_steps):

    # -----------------------------
    # Cars entering during this timestep
    # -----------------------------
    first_car = 0 if ts == 0 else Nend[ts - 1] + 1
    last_car = Nend[ts]

    if last_car >= first_car:
        for car in range(first_car, last_car + 1):

            dt = (ts + 1) * Delta - all_arrivals[car]
            start = all_start[car]
            entry_seg = segment(start)

            if entry_seg == -1:
                # entry point itself is outside the modeled corridor
                x[car][ts] = np.nan
                carloc[car][ts] = -1
                continue

            # entering cars used to ignore current traffic density
            # entirely (pure kinematics as if the road were empty). Now
            # they are slowed by the same density-based speed field as
            # cars already on the road, using the previous timestep's
            # density at their entry segment (or free-flow speed at
            # ts=0, when no density history exists yet).
            vrho_prev = vmax if ts == 0 else vrhol[ts - 1][entry_seg]

            x[car][ts], v[car][ts] = update_car(dt, a, start, 0.0, vrho_prev)
            carloc[car][ts] = segment(x[car][ts])

            if carloc[car][ts] != -1:
                count[ts][carloc[car][ts]] += 1

            loc[ts].append(x[car][ts])

    # -----------------------------
    # Cars already on the expressway
    # -----------------------------
    if ts > 0:

        for car in range(Nend[ts-1] + 1):

            seg = carloc[car][ts-1]

            if seg == -1:
               # already exited (or never validly entered) -- stays gone
                x[car][ts] = np.nan
                carloc[car][ts] = -1
                continue

            vrho_prev = vrhol[ts - 1][seg]

            x[car][ts], v[car][ts] = update_car(
                Delta, a, x[car][ts - 1], v[car][ts - 1], vrho_prev
            )

            carloc[car][ts] = segment(x[car][ts])

            if carloc[car][ts] != -1:
                count[ts][carloc[car][ts]] += 1

            loc[ts].append(x[car][ts])

    density[ts], vrhol[ts] = compute_density(count[ts])
    xloc[ts] = [row[ts] for row in x]

    active = sum(1 for c in carloc if c[ts] != -1)
    avg_density = float(np.mean(density[ts])) if num_segments else 0.0
    print(f"ts={ts:3d} | active cars={active:4d} | avg density={avg_density:.2f}")

    if verbose:
        print("--------------")
        for car in range(min(20, num_cars)):
            print(f"Car {car:3d} | Start={all_start[car]:5.0f} | "
                  f"x={x[car][ts]:8.2f} | Segment={carloc[car][ts]}")
```

```python
import matplotlib.pyplot as plt

# Define the number of cars to plot to avoid clutter
cars_to_plot = 20 # You can change this number

plt.figure(figsize=(15, 8))

# Plot the trajectory for each selected car
for car_idx in range(min(num_cars, cars_to_plot)):
    car_trajectory = [x[car_idx][ts] for ts in range(T_steps)]
    plt.plot(range(T_steps), car_trajectory, label=f'Car {car_idx+1}')

plt.xlabel('Time Step')
plt.ylabel('Car Location (meters)')
plt.title(f'Car Locations Over Time (First {min(num_cars, cars_to_plot)} Cars)')
plt.grid(True)
# Only show legend if the number of plotted cars is manageable
if cars_to_plot <= 10:
    plt.legend()

plt.show()
```

```python
def density_to_rgb(d):
    """White = no cars, green = free-flowing, red = at/over capacity."""
    if d <= 0:
        return (1.0, 1.0, 1.0)
    d = min(d, 1.0)  # clip anything over 100% capacity to full red
    if d < 0.5:
        t = d / 0.5
        return (t, 1.0, 0.0)          # green -> yellow
    else:
        t = (d - 0.5) / 0.5
        return (1.0, 1.0 - t, 0.0)    # yellow -> red
```

```python
def plot_segment_strip(ts, ax=None, show_legend=True):
    """Draw NLEX as a 1 x num_segments strip, colored by density at timestep ts."""
    row = np.array([density_to_rgb(d) for d in density[ts]]).reshape(1, num_segments, 3)

    if ax is None:
        fig, ax = plt.subplots(figsize=(18, 1.2))

    ax.imshow(row, aspect='auto', extent=[0, num_segments, 0, 1])
    ax.set_yticks([])
    ax.set_xlabel('Segment')
    ax.set_title(f'NLEX Congestion -- Timestep {ts} ({ts * Delta}s)')

    if show_legend:
        for label, d in [('No cars', 0), ('Free-flow', 0.1), ('Moderate', 0.5), ('Congested', 1.0)]:
            ax.scatter([], [], color=density_to_rgb(d), label=label, s=100, marker='s')
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.6), ncol=4, frameon=False)

    return ax

# Show the strip at the final timestep
plot_segment_strip(T_steps - 1)
plt.show()
```

```python
timesteps_to_show = list(range(0, T_steps, max(1, T_steps // 8)))

fig, axes = plt.subplots(len(timesteps_to_show), 1, figsize=(18, 1.0 * len(timesteps_to_show)))
if len(timesteps_to_show) == 1:
    axes = [axes]

for ax, ts in zip(axes, timesteps_to_show):
    plot_segment_strip(ts, ax=ax, show_legend=False)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

# -------------------------------
# Convert density to numpy array
# -------------------------------
density_array = np.array(density)

# -------------------------------
# Custom colormap
# White -> Green -> Red
# -------------------------------
colors = [
    (1, 1, 1),      # White
    (0, 0.8, 0),    # Green
    (1, 0, 0)       # Red
]

cmap = LinearSegmentedColormap.from_list(
    "traffic_density",
    colors,
    N=256
)

# -------------------------------
# Plot Heatmap
# -------------------------------
plt.figure(figsize=(18,6))

plt.imshow(
    density_array,
    origin='lower',
    aspect='auto',
    cmap=cmap,
    interpolation='nearest',
    vmin=0,
    vmax=1
)

plt.colorbar(label="Density")

plt.xlabel("Road Segment")
plt.ylabel("Time Step")

plt.title("Traffic Density Heatmap")

plt.xticks(
    ticks=np.arange(num_segments),
    labels=S["Exit_Distance"],
    rotation=90,
    fontsize=6
)

plt.yticks(np.arange(T_steps))

plt.tight_layout()
plt.show()
```
