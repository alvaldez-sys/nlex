```python
  import numpy as np

  # Parameters
  lam = 0.2 #cars per second
  T_max = 360 #in sec

  big_list = []

  def uniform_sequence(seed):
      rng = np.random.default_rng(seed)

      while True:
          yield rng.random()

  # Initialize generator
  for seed in range(1,17 + 1):
    # print(f'Seed: {seed}')
    gen = uniform_sequence(seed=seed)

    # Initial values
    i = 0
    t = [0.0]     # List of event times
    tau = []      # List of interarrival times

    # Run simulation until t > 3600
    while t[i] < T_max:

        # Generate Uniform(0,1)
        y = next(gen)

        # Generate exponential waiting time
        tau_i = -np.log(1 - y) / lam

        # Store waiting time
        tau.append(tau_i)

        # Compute next event time
        t_next = t[i] + tau_i
        t.append(t_next)

        # Print results
        #print(f"Event {i+1}")
        #print(f"rand = {y:.6f}")
        #print(f"tau = {tau_i:.6f}")
        #print(f"(t[{i}], seed={seed})    = {t_next:.6f}")
        big_list.append((float(round(t_next, 6)), seed))
        #print("-" * 30)

        # Move to next event
        i += 1

    print("Simulation finished.")
    print(f"Total events generated: {i}")
    print(f"Final time: {t[i]:.6f}")
    print()
    #Move to next index
    i += 1

  big_list.sort()
  print("List =", len(big_list))
  arrival =[]
  for entry in big_list:
    # print(f'({entry[0]:.6f}, {entry[1]})')
    arrival.append(entry[0])

  print('---')
  print(arrival)

  Nend=[]
  for upper_bound in range(5, round(max(t[0] for t in big_list)), 5):
    l2 = map(lambda f: (f[1][0], f[0]), enumerate(big_list))
    m = max(filter(lambda x: x[0] <= upper_bound, l2))
    # print(m)
    Nend.append(m[1])
  #print(Nend)


  Delta = 5
  ts = 0
  a = 0.0925

  max_cars = len(big_list)
  num_timesteps = 16
  #print(ts)
  x = [[0.0 for _ in range(num_timesteps)] for _ in range(max_cars)]
  v = [[0.0 for _ in range(num_timesteps)] for _ in range(max_cars)]

  print(Nend[ts] + 1)
  for car in range(0, Nend[ts]+1):
      dt = (ts + 1)*Delta - arrival[car]

      x[car][ts] = 0.5 * a * dt**2

      if dt != 0:
          v[car][ts] = x[car][ts] / dt
      else:
          v[car][ts] = 0.0
      #print(f"x[{car}, {ts}] = {x[car][ts]}")
      #print(f"v[{car}, {ts}] = {v[car][ts]}")


  while ts < 15:
    ts += 1
    print('-------')
    for car in range(Nend[ts -1]+1, Nend[ts]+1):
      dt = (ts + 1)*Delta - arrival[car]

      x[car][ts] = 0.5 * a * dt**2

      if dt != 0:
          v[car][ts] = x[car][ts] / dt
      else:
          v[car][ts] = 0.0
      #print(f"x[{car}, {ts}] = {x[car][ts]}")
      #print(f"v[{car}, {ts}] = {v[car][ts]}")

    for car in range(Nend[ts - 1]+1):
      dt = Delta
      x[car][ts] = x[car][ts - 1] + v[car][ts - 1]*dt + 0.5 * a * dt**2
      v[car][ts] = (x[car][ts] - x[car][ts - 1]) / dt
      #print(f"x[{car}, {ts}] = {x[car][ts]}")
      #print(f"v[{car}, {ts}] = {v[car][ts]}")


  for ts in range(16):
    #for car in range(Nend[ts]+1):
      print(f"x[{0}, {ts}] = {x[0][ts]}")
      #print(f"v[{car}, {ts}] = {v[car][ts]}")
      print('___')


```
